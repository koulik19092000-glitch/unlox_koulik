# Section A — Theory

**Q1. `is` vs `==`**

`==` compares **values** (are the contents equal?). `is` compares **identity** (are both names pointing to the *same object in memory*?).
Two separate lists with the same contents are equal but not identical.

````python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)   # True  -> same value
print(a is b)   # False -> two different objects in memory
````

**Q2. Mutable vs Immutable**

- **Mutable:** list, set, dict
- **Immutable:** tuple, str, int, float, bool, frozenset

*Why it matters:* immutable objects can't be changed by accident when shared across functions, and they can be used as dict keys / set members (they are hashable).

**Q3. `0.1 + 0.2 == 0.3`** → predicted output: `False`

Computers store floats in binary, and 0.1 / 0.2 can't be represented exactly (like 1/3 in decimal), so the sum is `0.30000000000000004`. Not a Python bug, it's IEEE-754. Compare with a tolerance instead.

````python
print(0.1 + 0.2 == 0.3)
print(0.1 + 0.2)
print(abs((0.1 + 0.2) - 0.3) < 1e-9)   # the correct way
````

**Q4. Tuple over list — 3 situations**

1. **Fixed records** like `(latitude, longitude)` or `(day, month, year)`: the size/meaning is fixed, and nobody should modify it.
2. **Dictionary keys / set members**, e.g. `{(12.97, 77.59): "Bengaluru"}`: tuples are hashable, lists aren't.
3. **Returning multiple values from a function**, or protecting constant data (like weekday names): immutability prevents accidental edits, and tuples are slightly faster and lighter than lists.

**Q5. Why dict lookup is O(1) on average but `in` on a list is O(n)**

*Without the word "hash":* a list has no idea where a value is, so it must check items one by one from the start. A dict turns the key into a number using a formula, and that number tells it exactly which slot to open, like a locker number, so it jumps straight there no matter how big the dict is.

*With the word "hash":* Python computes `hash(key)`, maps it to a bucket index (`hash % table_size`), and goes straight to that bucket. Collisions are rare, so the average cost is constant, O(1). A list `in` has to compare against each element, O(n).

**Q6. `range()` vs `numpy.arange()`**

1. `range` accepts **only integers**; `np.arange(0, 1, 0.1)` accepts **floats**.
2. `range` is a lazy sequence object; `arange` returns a real **ndarray** that supports vectorised maths (`arr * 2`).
3. `arange` has a `dtype` argument and is stored in contiguous typed memory; `range` has no dtype and stores nothing.
4. (Bonus) with float steps, `arange`'s length can be off by rounding errors; `range` is exact.

**Q7. `append` vs `extend` vs `insert`** with `a = [1, 2, 3]` and `x = [4, 5]`

````python
a = [1, 2, 3]; a.append([4, 5]);     print("append :", a)   # adds x as ONE element (nested list)
a = [1, 2, 3]; a.extend([4, 5]);     print("extend :", a)   # adds each item of x separately
a = [1, 2, 3]; a.insert(1, [4, 5]);  print("insert :", a)   # puts x as ONE element at index 1
````

**Q8. Star unpacking**

`a, *b, c = [10, 20, 30, 40, 50]` → first item goes to `a`, last to `c`, and everything in the middle becomes a **list** in `b`.

**Q10. Why NumPy beats Python lists**

1. **Memory layout:** a NumPy array stores same-type values in one contiguous block; a list stores pointers to separate objects scattered in memory. Contiguous data is cache-friendly and compact.
2. **Vectorised C loops:** the loop runs in compiled C, with no per-element type checking or interpreter overhead.
3. **Fixed dtype + SIMD:** since every element has the same type, the CPU can process many values per instruction.

**Q16. Swap without a third variable**

````python
def swap_demo(a, b):
    # (a) arithmetic
    x, y = a, b
    x = x + y
    y = x - y
    x = x - y
    print("Arithmetic :", x, y)

    # (b) XOR (integers only)
    x, y = a, b
    x = x ^ y
    y = x ^ y
    x = x ^ y
    print("XOR        :", x, y)

    # (c) tuple unpacking
    x, y = a, b
    x, y = y, x
    print("Unpacking  :", x, y)

swap_demo(10, 25)
# Most Pythonic: (c). It's one readable line, works for any type (not just numbers),
# and has no overflow/int-only limits. (a) and (b) are interview tricks.
````

**Q19. Sum of digits (no string conversion)**

````python
def digit_sum(n):
    n = abs(n)
    total = 0
    while n > 0:
        total += n % 10     # last digit
        n //= 10            # drop last digit
    return total

# n = int(input("Enter a 4-digit number: "))
print(digit_sum(4829))   # 4+8+2+9 = 23
````

**Q21. Income tax (New Regime, slab-wise)**

````python
def income_tax(income):
    # (upper limit of slab, rate %)
    slabs = [(300000, 0), (700000, 5), (1000000, 10),
             (1200000, 15), (1500000, 20), (float("inf"), 30)]
    lower = 0
    total_tax = 0
    print(f"Annual income: ₹{income:,}")
    for upper, rate in slabs:
        if income > lower:
            taxable = min(income, upper) - lower
            tax = taxable * rate / 100
            total_tax += tax
            upper_text = "above" if upper == float("inf") else f"to ₹{upper:,}"
            print(f"  ₹{lower:,} {upper_text:<14} @ {rate:>2}% on ₹{taxable:>9,} = ₹{tax:,.0f}")
        lower = upper
    print(f"Total tax: ₹{total_tax:,.0f}")

# income_tax(int(input("Annual income: ")))
income_tax(800000)    # expected 30,000
print()
income_tax(1800000)
````

**Q22. PAN validator (no regex)**

````python
def validate_pan(pan):
    if len(pan) != 10:
        return f"Invalid PAN — length must be 10, got {len(pan)}"
    for i in range(10):
        ch = pan[i]
        if i < 5 or i == 9:
            # positions 1-5 and 10 must be uppercase letters
            if not ("A" <= ch <= "Z"):
                return f"Invalid PAN — character {i + 1} ('{ch}') must be an uppercase letter"
        else:
            if not ("0" <= ch <= "9"):
                return f"Invalid PAN — character {i + 1} ('{ch}') must be a digit"
    return "Valid PAN"

# print(validate_pan(input("PAN: ")))
for p in ["ABCDE1234F", "abcde1234f", "ABCD12345F", "ABCDE12345", "ABC1234"]:
    print(f"{p:<12} -> {validate_pan(p)}")
````

**Q23. BESCOM electricity bill**

````python
def bescom_bill(units):
    # (units in this slab, rate) — slabs are applied progressively
    slabs = [(50, 4.10), (50, 5.55), (100, 7.10), (float("inf"), 8.15)]
    remaining = units
    energy = 0
    print(f"Units consumed: {units}")
    for slab_units, rate in slabs:
        if remaining <= 0:
            break
        used = min(remaining, slab_units)
        cost = used * rate
        energy += cost
        print(f"  {used:>5} units x ₹{rate:.2f} = ₹{cost:>8.2f}")
        remaining -= used
    fixed = 110
    tax = 0.06 * (energy + fixed)
    print(f"Energy charges : ₹{energy:.2f}")
    print(f"Fixed charge   : ₹{fixed:.2f}")
    print(f"Tax (6%)       : ₹{tax:.2f}")
    print(f"Total bill     : ₹{energy + fixed + tax:.2f}")

# bescom_bill(float(input("Units: ")))
bescom_bill(230)
````

**Q24. Cricket strike-rate classifier**

````python
def classify_strike_rate(runs, balls):
    if balls == 0:
        print("Cannot compute strike rate: balls faced is 0.")
        return
    sr = runs / balls * 100
    if sr > 150:
        label = "Explosive (T20 finisher)"
    elif sr > 100:
        label = "Aggressive (Top order T20)"
    elif sr >= 75:
        label = "Steady (ODI anchor)"
    else:
        label = "Defensive (Test match)"
    print(f"Strike rate = {sr:.2f} -> {label}")

# classify_strike_rate(int(input("Runs: ")), int(input("Balls: ")))
for r, b in [(60, 30), (45, 35), (80, 90), (20, 50), (10, 0)]:
    classify_strike_rate(r, b)
````

**Q27. Number diamond**

````python
def diamond_row(i, n):
    for _ in range(n - i):          # each element is 2 chars wide -> indent 2 per missing level
        print(" ", end=" ")
    for j in range(1, i + 1):       # going up
        print(j, end=" ")
    for j in range(i - 1, 0, -1):   # coming back down
        print(j, end=" ")
    print()

N = 5
for i in range(1, N + 1):
    diamond_row(i, N)
for i in range(N - 1, 0, -1):
    diamond_row(i, N)
````

**Q31. Armstrong numbers (1–9999)**

````python
armstrong = []
for num in range(1, 10000):
    # count digits
    digits = 0
    temp = num
    while temp > 0:
        digits += 1
        temp //= 10
    # sum of digits ** digits
    total = 0
    temp = num
    while temp > 0:
        total += (temp % 10) ** digits
        temp //= 10
    if total == num:
        armstrong.append(num)

print(", ".join(str(x) for x in armstrong))
print("Count:", len(armstrong))
````

**Q32. Fibonacci (iterative)**

````python
def fibonacci(n):
    a, b = 0, 1
    terms = []
    total = 0
    for _ in range(n):
        terms.append(a)
        total += a
        a, b = b, a + b
    print(" ".join(str(t) for t in terms))
    print("Sum =", total)

# fibonacci(int(input("N: ")))
for n in [0, 1, 10]:
    print(f"N = {n}")
    fibonacci(n)
````

**Q33. Optimised prime check + sum of primes**

````python
def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:          # only up to sqrt(n)
        if n % i == 0:
            return False
        i += 1
    return True

def prime_report(n):
    print(f"{n} is prime" if is_prime(n) else f"{n} is not prime")
    total = 0
    for k in range(2, n + 1):
        if is_prime(k):
            total += k
    print(f"Sum of primes from 2 to {n} = {total}")

# prime_report(int(input("N: ")))
prime_report(29)
prime_report(10)
````

**Q34. Reverse a number / palindrome (no strings)**

````python
def reverse_number(n):
    rev = 0
    while n > 0:
        rev = rev * 10 + n % 10
        n //= 10
    return rev

def check_palindrome(n):
    if n < 0:
        print(f"{n} -> not palindrome (negative)")
        return
    rev = reverse_number(n)
    print(f"{n} reversed = {rev} -> " + ("palindrome" if rev == n else "not palindrome"))

# check_palindrome(int(input("Number: ")))
for n in [121, 123, -121, 0, 1221]:
    check_palindrome(n)
````

**Q35. HCF (Euclid) and LCM**

````python
def hcf_lcm(a, b):
    x, y = a, b
    while y != 0:
        x, y = y, x % y
    hcf = x
    lcm = a * b // hcf
    print(f"HCF({a}, {b}) = {hcf}, LCM = {lcm}")

# hcf_lcm(int(input()), int(input()))
hcf_lcm(48, 18)
hcf_lcm(17, 5)
````

**Q36. String scanner (single pass)**

````python
def scan_string(s):
    total = words = vowels = consonants = digits = spaces = special = 0
    in_word = False
    for ch in s:
        total += 1
        if ch == " ":
            spaces += 1
            in_word = False
        else:
            if not in_word:         # a new word starts at a non-space after a space
                words += 1
                in_word = True
            if ch.isdigit():
                digits += 1
            elif ch.isalpha():
                if ch.lower() in "aeiou":
                    vowels += 1
                else:
                    consonants += 1
            else:
                special += 1
    print(f"Characters: {total}, Words: {words}, Vowels: {vowels}, Consonants: {consonants}")
    print(f"Digits: {digits}, Spaces: {spaces}, Special: {special}")

# scan_string(input("Sentence: "))
scan_string("Order came on time, food was hot, packaging 10/10!")
````

**Q37. Marks analysis (no sort)**

````python
def analyse_marks(line):
    marks = list(map(int, line.split()))
    highest = lowest = marks[0]
    total = 0
    for m in marks:
        total += m
        if m > highest:
            highest = m
        if m < lowest:
            lowest = m
    average = total / len(marks)

    above_avg = 0
    second = None                  # second highest DISTINCT mark
    for m in marks:
        if m > average:
            above_avg += 1
        if m < highest and (second is None or m > second):
            second = m

    print("Marks:", marks)
    print("(a) Maximum:", highest)
    print("(b) Minimum:", lowest)
    print("(c) Average:", average)
    print("(d) Above average:", above_avg)
    print("(e) Second highest:", second)

# analyse_marks(input("Enter 10 marks: "))
analyse_marks("78 92 65 88 92 45 71 99 54 83")
````

**Q38. Word frequency analyser**

````python
def word_frequency(review):
    # lowercase and turn every non-alphanumeric character into a space
    cleaned = ""
    for ch in review.lower():
        cleaned += ch if ch.isalnum() else " "

    freq = {}
    for word in cleaned.split():
        freq[word] = freq.get(word, 0) + 1

    # simple bubble sort by count, descending (strict < keeps ties in first-seen order)
    items = list(freq.items())
    for i in range(len(items)):
        for j in range(len(items) - 1 - i):
            if items[j][1] < items[j + 1][1]:
                items[j], items[j + 1] = items[j + 1], items[j]

    for word, count in items:
        print(f"{word:<10} {count}")
    print("Most frequent:", items[0][0], "->", items[0][1])

# word_frequency(input("Review: "))
word_frequency("Food was great, FOOD was hot and the delivery was fast. Great food!")
````

**Q43. Smart palindrome checker**

````python
def is_smart_palindrome(sentence):
    cleaned = ""
    for ch in sentence:
        if ch.isalnum():
            cleaned += ch.lower()
    left, right = 0, len(cleaned) - 1
    while left < right:
        if cleaned[left] != cleaned[right]:
            return False
        left += 1
        right -= 1
    return True

for s in ["A man, a plan, a canal: Panama", "Race car", "No 'x' in Nixon", "Hello world"]:
    print(f"{s!r:<35} -> {is_smart_palindrome(s)}")
````

````python
def manual_title(sentence):
    result_words = []
    for word in sentence.split():
        result_words.append(word[0].upper() + word[1:].lower())
    return " ".join(result_words)

print(manual_title("hello world from bengaluru"))
print(manual_title("pYTHON iS fUN"))
````

**Q45. Caesar cipher**

````python
def caesar_encrypt(message, shift):
    result = ""
    for ch in message:
        if "A" <= ch <= "Z":
            result += chr((ord(ch) - ord("A") + shift) % 26 + ord("A"))
        elif "a" <= ch <= "z":
            result += chr((ord(ch) - ord("a") + shift) % 26 + ord("a"))
        else:
            result += ch          # digits, spaces, symbols unchanged
    return result

print(caesar_encrypt("Hello, India!", 3))     # Khoor, Lqgld!
print(caesar_encrypt("xyz ABC 123", 3))       # wrap-around check
````

## C7 — Functions

**Q46. `calculate_gst`**

````python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if gst_inclusive:
        base = amount / (1 + gst_rate / 100)   # amount already contains GST
        gst = amount - base
        total = amount
    else:
        base = amount
        gst = amount * gst_rate / 100
        total = base + gst
    return (round(base, 2), round(gst, 2), round(total, 2))

print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
````

**Q47. Recursive `power`**

````python
def power(base, exp):
    if exp == 0:
        return 1
    if exp < 0:
        return 1 / power(base, -exp)
    return base * power(base, exp - 1)

print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
````

**Q48. Identity matrix manipulations**

````python
import numpy as np

identity = np.eye(5)
print("(a) Identity 5x5:\n", identity)

scaled = identity * 10
print("(b) Multiplied by 10:\n", scaled)

# (c) Transpose swaps rows and columns, i.e. element [i][j] <-> [j][i].
# In an identity matrix the only non-zero elements are on the diagonal (i == j),
# which don't move, and everything else is 0, so the matrix is symmetric: A.T == A.
transposed = scaled.T
print("(c) Transpose:\n", transposed)
print("    Same as original?", np.array_equal(scaled, transposed))

print("(d) Sum of diagonal:", np.trace(scaled))
print("(e) Sum of all elements:", scaled.sum())
````
